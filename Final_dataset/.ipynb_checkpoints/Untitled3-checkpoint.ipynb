```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import numpy as np
import os

# ─── STYLE (IDENTIQUE) ────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "serif",
    "font.serif":         ["Times New Roman", "DejaVu Serif"],
    "font.size":          10,
    "axes.titlesize":     10,
    "axes.labelsize":     9,
    "xtick.labelsize":    8,
    "ytick.labelsize":    8,
    "legend.fontsize":    7,
    "legend.framealpha":  0.85,
    "legend.edgecolor":   "0.7",
    "axes.linewidth":     0.8,
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "grid.linewidth":     0.4,
    "grid.alpha":         0.4,
    "grid.color":         "0.75",
    "lines.linewidth":    1.2,
    "pdf.fonttype":       42,
    "ps.fonttype":        42,
})

sns.set_style("whitegrid")

# ─── PALETTE EXACTE ───────────────────────────────────────────────────
COLORS    = ["#0072B2", "#E69F00"]
LINESTYLE = ["-", "--"]

# ─── CONFIG ───────────────────────────────────────────────────────────
DELAYS = ["2ms", "10ms", "50ms", "100ms", "35s"]

FILES = {
    "Local":  {d: f"scheduler_times-{d}.csv"        for d in DELAYS},
    "Hybrid": {d: f"scheduler_times-{d}_edited.csv" for d in DELAYS},
}

MODES = ["Local", "Hybrid"]

# ─── LOAD ─────────────────────────────────────────────────────────────
def load_series(filepath):
    if not os.path.exists(filepath):
        return None

    with open(filepath, "r") as f:
        first_line = f.readline()

    sep = ";" if ";" in first_line else ","
    df = pd.read_csv(filepath, sep=sep)

    try:
        pd.to_numeric(df.columns[0])
        df = pd.read_csv(filepath, sep=sep, header=None)
    except:
        pass

    numeric_cols = df.select_dtypes(include=[np.number]).columns
    return df[numeric_cols[0]].dropna()

# ─── STYLE AXES ───────────────────────────────────────────────────────
def style_ax(ax):
    ax.yaxis.set_major_locator(ticker.MaxNLocator(nbins=5, prune='both'))
    ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=5, prune='both'))
    ax.tick_params(direction='out', length=3, width=0.6, pad=3)
    ax.grid(True, linestyle=':', linewidth=0.4, color='0.75')

# ─── FIGURE ───────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 5, figsize=(18, 4), sharey=True)

# ─── PLOT ─────────────────────────────────────────────────────────────
for c, delay in enumerate(DELAYS):
    ax = axes[c]

    for i, mode in enumerate(MODES):
        data = load_series(FILES[mode][delay])
        if data is None:
            continue

        sns.ecdfplot(
            data=data,
            ax=ax,
            color=COLORS[i],
            linestyle=LINESTYLE[i],
            linewidth=1.2,
            label=mode
        )

        # percentiles
        p50 = np.percentile(data, 50)
        p95 = np.percentile(data, 95)

        ax.axvline(p50, color=COLORS[i], linestyle='-', linewidth=1.0, alpha=0.6)
        ax.axvline(p95, color=COLORS[i], linestyle=':', linewidth=1.0, alpha=0.8)

    ax.set_title(delay, fontweight='bold')
    ax.set_xlabel("Latency (ms)")

    if c == 0:
        ax.set_ylabel("CDF")

    ax.set_ylim(0, 1.05)
    ax.yaxis.set_major_locator(ticker.MultipleLocator(0.25))
    ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda v, _: f"{v:.2f}"))

    ax.legend(
        loc='lower right',
        handlelength=1.8,
        borderpad=0.4,
        labelspacing=0.2,
        handletextpad=0.4
    )

    style_ax(ax)

# ─── TITRE ────────────────────────────────────────────────────────────
fig.suptitle(
    "Scheduler Latency — Local vs Hybrid (Cold & Warm)",
    fontsize=12,
    fontweight='bold',
    y=0.98
)

# ─── LAYOUT ───────────────────────────────────────────────────────────
plt.tight_layout()

# ─── EXPORT ───────────────────────────────────────────────────────────
plt.savefig("scheduler_ecdf_overlay_5cols.pdf", dpi=300, bbox_inches="tight")
plt.savefig("scheduler_ecdf_overlay_5cols.png", dpi=300, bbox_inches="tight")

plt.show()
print("✓ Saved: scheduler_ecdf_overlay_5cols.pdf / .png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import numpy as np
import os

# ─── STYLE ────────────────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "serif",
    "font.serif":         ["Times New Roman", "DejaVu Serif"],
    "font.size":          10,
    "axes.titlesize":     10,
    "axes.labelsize":     9,
    "xtick.labelsize":    8,
    "ytick.labelsize":    8,
    "legend.fontsize":    7,
    "legend.framealpha":  0.85,
    "legend.edgecolor":   "0.7",
    "axes.linewidth":     0.8,
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "grid.linewidth":     0.4,
    "grid.alpha":         0.4,
    "grid.color":         "0.75",
    "lines.linewidth":    1.2,
    "pdf.fonttype":       42,
    "ps.fonttype":        42,
})

sns.set_style("whitegrid")

# ─── PALETTE ──────────────────────────────────────────────────────────
COLORS     = ["#0072B2", "#E69F00", "#009E73"]   # bleu, orange, vert
LINESTYLES = ["-",       "--",      "-."]
LABELS     = ["Local", "Hybrid", "Local 1core"]

# ─── CONFIG ───────────────────────────────────────────────────────────
DELAYS = ["2ms", "10ms", "50ms", "100ms", "35s"]

# Fichiers par mode et délai — tous ont le même format (timestamp,pod,scheduling_time_ms)
FILES = {
    "Local": {
        d: f"scheduler_times-{d}.csv" for d in DELAYS
    },
    "Hybrid": {
        d: f"scheduler_times-{d}_edited.csv" for d in DELAYS
    },
    "Local 1core": {
        d: f"scheduler_times-{d}_1core.csv" for d in DELAYS
    },
}

MODES = ["Local", "Hybrid", "Local 1core"]

# ─── CHARGEMENT ───────────────────────────────────────────────────────
def load_series(filepath):
    """Charge la colonne scheduling_time_ms depuis un CSV (timestamp,pod,scheduling_time_ms)."""
    if filepath is None or not os.path.exists(filepath):
        return None
    df = pd.read_csv(filepath)
    return df["scheduling_time_ms"].dropna()

# ─── STYLE AXES ───────────────────────────────────────────────────────
def style_ax(ax):
    ax.yaxis.set_major_locator(ticker.MaxNLocator(nbins=5, prune='both'))
    ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=5, prune='both'))
    ax.tick_params(direction='out', length=3, width=0.6, pad=3)
    ax.grid(True, linestyle=':', linewidth=0.4, color='0.75')

# ─── FIGURE ───────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 5, figsize=(18, 4), sharey=True)

for c, delay in enumerate(DELAYS):
    ax = axes[c]

    for i, mode in enumerate(MODES):
        filepath = FILES[mode][delay]
        data = load_series(filepath)
        if data is None:
            continue

        sns.ecdfplot(
            data=data,
            ax=ax,
            color=COLORS[i],
            linestyle=LINESTYLES[i],
            linewidth=1.2,
            label=LABELS[i],
        )



    ax.set_title(delay, fontweight='bold')
    ax.set_xlabel("Latency (ms)")

    if c == 0:
        ax.set_ylabel("CDF")

    ax.set_ylim(0, 1.05)
    ax.yaxis.set_major_locator(ticker.MultipleLocator(0.25))
    ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda v, _: f"{v:.2f}"))

    ax.legend(
        loc='lower right',
        handlelength=1.8,
        borderpad=0.4,
        labelspacing=0.2,
        handletextpad=0.4,
    )

    style_ax(ax)

# ─── TITRE ────────────────────────────────────────────────────────────
fig.suptitle(
    "Scheduler Latency — Local vs Hybrid vs Local 1core ",
    fontsize=12,
    fontweight='bold',
    y=0.98,
)

plt.tight_layout()

# ─── EXPORT ───────────────────────────────────────────────────────────
plt.savefig("scheduler_ecdf_overlay_5cols.pdf", dpi=300, bbox_inches="tight")
plt.savefig("scheduler_ecdf_overlay_5cols.png", dpi=300, bbox_inches="tight")

plt.show()
print("✓ Saved: scheduler_ecdf_overlay_5cols.pdf / .png")
```
