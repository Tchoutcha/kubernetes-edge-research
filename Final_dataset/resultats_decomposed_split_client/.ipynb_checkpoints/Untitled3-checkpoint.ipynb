```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

# ── Configuration ──────────────────────────────────────────────────────────────
DELAYS  = ["2ms", "50ms", "100ms"]
PERCS   = [50, 90, 95, 99]
PERC_LABELS = ["p50", "p90", "p95", "p99"]

FILES = {
    "original": {
        d: {
            "all_data": os.path.join("all_data", f"all_data_{d}.csv"),
            "kourier":  os.path.join("kourier",  f"kourier_{d}.csv"),
        }
        for d in DELAYS
    },
    "edited": {
        d: {
            "all_data": os.path.join("all_data", f"all_data_{d}_edited.csv"),
            "kourier":  os.path.join("kourier",  f"kourier_{d}_edited.csv"),
        }
        for d in DELAYS
    },
}

# Colonnes
COL_T1 = {"original": "time_taken_s", "edited": "latency_s"}  # déjà en ms
COL_T2 = "duration_ms"

# ── Chargement ─────────────────────────────────────────────────────────────────
def read_col(filepath: str, col: str) -> pd.Series:
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Fichier introuvable : {filepath}")
    with open(filepath, "r") as f:
        first_line = f.readline()
    sep = ";" if ";" in first_line else ","
    df = pd.read_csv(filepath, sep=sep)
    if col not in df.columns:
        raise ValueError(f"Colonne '{col}' absente dans {filepath}. Colonnes : {list(df.columns)}")
    return df[col].dropna()

# ── Calcul overhead par percentile (approche statistique) ─────────────────────
rows = []
for version in ["original", "edited"]:
    for delay in DELAYS:
        try:
            t1 = read_col(FILES[version][delay]["all_data"], COL_T1[version])
            t2 = read_col(FILES[version][delay]["kourier"],  COL_T2)

            row = {
                "Version": version.capitalize(),
                "Délai":   delay,
                "n_t1":    len(t1),
                "n_t2":    len(t2),
            }
            for p, label in zip(PERCS, PERC_LABELS):
                p_t1 = np.percentile(t1, p)
                p_t2 = np.percentile(t2, p)
                row[f"{label}_t1"]       = round(p_t1, 2)
                row[f"{label}_t2"]       = round(p_t2, 2)
                row[f"{label}_overhead"] = round(p_t1 - p_t2, 2)

            rows.append(row)
            print(f"\n{'='*50}")
            print(f"  {version.upper()} — Délai {delay}")
            print(f"{'='*50}")
            print(f"  {'Percentile':<12} {'t1 (ms)':<12} {'t2 (ms)':<12} {'Overhead (ms)'}")
            print(f"  {'-'*50}")
            for p, label in zip(PERCS, PERC_LABELS):
                print(f"  {label:<12} {row[f'{label}_t1']:<12} {row[f'{label}_t2']:<12} {row[f'{label}_overhead']}")

        except (FileNotFoundError, ValueError) as e:
            print(f"⚠️  {e}")

df_perc = pd.DataFrame(rows)

# ── Figure : tableau matplotlib ────────────────────────────────────────────────
col_labels = ["Version", "Délai", "n",
              "p50 t1", "p50 t2", "p50 Δ",
              "p90 t1", "p90 t2", "p90 Δ",
              "p95 t1", "p95 t2", "p95 Δ",
              "p99 t1", "p99 t2", "p99 Δ"]

cell_data = []
for _, r in df_perc.iterrows():
    cell_data.append([
        r["Version"], r["Délai"], int(r["n_t1"]),
        f'{r["p50_t1"]:.1f}', f'{r["p50_t2"]:.1f}', f'{r["p50_overhead"]:.1f}',
        f'{r["p90_t1"]:.1f}', f'{r["p90_t2"]:.1f}', f'{r["p90_overhead"]:.1f}',
        f'{r["p95_t1"]:.1f}', f'{r["p95_t2"]:.1f}', f'{r["p95_overhead"]:.1f}',
        f'{r["p99_t1"]:.1f}', f'{r["p99_t2"]:.1f}', f'{r["p99_overhead"]:.1f}',
    ])

fig, ax = plt.subplots(figsize=(18, 4))
fig.patch.set_facecolor("white")
ax.axis("off")

table = ax.table(
    cellText=cell_data,
    colLabels=col_labels,
    cellLoc="center",
    loc="center",
)
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.1, 1.8)

HEADER_COLOR = "#2C3E50"
ORIG_COLOR   = "#EBF3FF"
EDIT_COLOR   = "#EDFBF0"
DELTA_COLOR  = "#FFF3E0"  # orange clair pour les colonnes overhead Δ

# Indices des colonnes Δ (overhead) : 5, 8, 11, 14
DELTA_COLS = [5, 8, 11, 14]

for (row_idx, col_idx), cell in table.get_celld().items():
    cell.set_edgecolor("#CCCCCC")
    cell.set_linewidth(0.5)
    if row_idx == 0:
        cell.set_facecolor(HEADER_COLOR)
        cell.set_text_props(color="white", fontweight="bold", fontsize=9)
    else:
        version = cell_data[row_idx - 1][0]
        if col_idx in DELTA_COLS:
            cell.set_facecolor(DELTA_COLOR)
            cell.set_text_props(fontweight="bold")
        else:
            cell.set_facecolor(ORIG_COLOR if version == "Original" else EDIT_COLOR)

fig.suptitle(
    "Overhead réseau par percentile (Δ = t1 − t2) — approche statistique — Knative / k3s sur Raspberry Pi",
    fontsize=12, fontweight="bold", color="#111111", y=0.98,
)
fig.text(
    0.5, 0.01,
    "t1 = latence client (ms)  |  t2 = duration Kourier (ms)  |  Δ = overhead Knative/scheduling (ms)  |  Colonnes orange = overhead",
    ha="center", fontsize=8, color="#555555", style="italic",
)

plt.savefig("percentiles_overhead.png", dpi=300, bbox_inches="tight", facecolor="white")
print("\n✅ Tableau sauvegardé : percentiles_overhead.png")

df_perc.to_csv("percentiles_overhead.csv", index=False)
print("✅ CSV exporté : percentiles_overhead.csv")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import os

# ─── Style publication ASPLOS/OSDI ────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "serif",
    "font.serif":         ["Times New Roman", "DejaVu Serif"],
    "font.size":          10,
    "axes.titlesize":     10,
    "axes.labelsize":     9,
    "xtick.labelsize":    8,
    "ytick.labelsize":    8,
    "legend.fontsize":    8,
    "legend.framealpha":  0.85,
    "legend.edgecolor":   "0.7",
    "axes.linewidth":     0.8,
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "grid.linewidth":     0.4,
    "grid.alpha":         0.4,
    "grid.color":         "0.75",
    "lines.linewidth":    1.4,
    "pdf.fonttype":       42,
    "ps.fonttype":        42,
})

# ─── Configuration ────────────────────────────────────────────────────────────
DELAYS      = ["2ms", "10ms", "50ms", "100ms"]
PERCS       = [50, 90, 95, 99]
PERC_LABELS = ["p50", "p90", "p95", "p99"]

FILES = {
    "original": {
        d: {
            "all_data": os.path.join("all_data", f"all_data_{d}.csv"),
            "kourier":  os.path.join("kourier",  f"kourier_{d}.csv"),
        }
        for d in DELAYS
    },
    "edited": {
        d: {
            "all_data": os.path.join("all_data", f"all_data_{d}_edited.csv"),
            "kourier":  os.path.join("kourier",  f"kourier_{d}_edited.csv"),
        }
        for d in DELAYS
    },
}

COL_T1 = {"original": "latency_s", "edited": "latency_s"}
COL_T2 = "duration_ms"

VERSION_COLORS  = {"original": "#0072B2", "edited": "#D55E00"}
VERSION_LABELS  = {"original": "Local",   "edited": "Hybrid"}

# ─── Chargement ───────────────────────────────────────────────────────────────
def read_col(filepath, col):
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"File not found: {filepath}")
    with open(filepath, "r") as f:
        sep = ";" if ";" in f.readline() else ","
    df = pd.read_csv(filepath, sep=sep)
    if col not in df.columns:
        raise ValueError(f"Column '{col}' not found in {filepath}")
    return df[col].dropna()

# ─── Calcul overhead Δ par percentile ─────────────────────────────────────────
overhead = {"original": {}, "edited": {}}

for version in ["original", "edited"]:
    for delay in DELAYS:
        try:
            t1 = read_col(FILES[version][delay]["all_data"], COL_T1[version])
            t2 = read_col(FILES[version][delay]["kourier"],  COL_T2)
            overhead[version][delay] = [
                np.percentile(t1, p) - np.percentile(t2, p) for p in PERCS
            ]
        except (FileNotFoundError, ValueError) as e:
            print(f"⚠  {e}")
            overhead[version][delay] = [np.nan] * len(PERCS)

# ─── Helpers ──────────────────────────────────────────────────────────────────
def style_ax(ax):
    ax.yaxis.set_major_locator(ticker.MaxNLocator(nbins=5, prune='both'))
    ax.tick_params(direction='out', length=3, width=0.6, pad=3)
    ax.grid(True, axis='y', linestyle=':', linewidth=0.4, color='0.75')
    ax.axhline(0, color='0.3', linewidth=0.8, linestyle='--')

# ─── Figure 1 × 4 ─────────────────────────────────────────────────────────────
fig, axes = plt.subplots(
    1, len(DELAYS),
    figsize=(15, 4),
    sharey=False,
    constrained_layout=True,
)

x       = np.arange(len(PERC_LABELS))
width   = 0.30
offsets = [-width / 2, width / 2]

for col_idx, delay in enumerate(DELAYS):
    ax = axes[col_idx]

    for i, version in enumerate(["original", "edited"]):
        deltas = overhead[version][delay]
        bars = ax.bar(
            x + offsets[i],
            deltas,
            width=width,
            color=VERSION_COLORS[version],
            alpha=0.85,
            label=VERSION_LABELS[version],
            edgecolor='white',
            linewidth=0.4,
        )
        # Annotations valeurs
        for bar, val in zip(bars, deltas):
            if not np.isnan(val):
                va = 'bottom' if val >= 0 else 'top'
                yy = bar.get_height() if val >= 0 else 0
                ax.text(
                    bar.get_x() + bar.get_width() / 2,
                    yy,
                    f"{val:.1f}",
                    ha='center', va=va,
                    fontsize=6.5, color='0.2',
                    fontfamily='monospace',
                )

    ax.set_xticks(x)
    ax.set_xticklabels(PERC_LABELS)
    ax.set_xlabel("Percentile")
    ax.set_ylabel("Overhead (ms)")
    ax.set_title(delay, fontsize=10, fontweight='bold', pad=6)
    ax.legend(loc='upper left', handlelength=1.5,
              borderpad=0.4, labelspacing=0.2, handletextpad=0.4)
    style_ax(ax)

# ─── Titre principal ──────────────────────────────────────────────────────────
fig.suptitle(
    "Overhead per Percentile — t1 (client latency) − t2 (Kourier duration)",
    fontsize=11, fontweight='bold',
)

# ─── Sous-titre explicatif ────────────────────────────────────────────────────
fig.text(
    0.5, -0.04,
    "t1 = client latency (ms)  |  t2 = Kourier duration (ms)",
    ha='center', fontsize=8, color='0.45', style='italic',
)

# ─── Export ───────────────────────────────────────────────────────────────────
plt.savefig("overhead_barchart.pdf", dpi=300, bbox_inches="tight")
plt.savefig("overhead_barchart.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ Saved: overhead_barchart.pdf / .png")
```
