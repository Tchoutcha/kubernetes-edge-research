```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
files = [
    "invocation_results_fixed1.csv",
    "invocation_results_fixed2.csv",
    "invocation_results_fixed3.csv",
    "invocation_results_fixed4.csv",
    "invocation_results_fixed5.csv",
    "invocation_results_fixed6.csv",
    "invocation_results_fixed7.csv",
    "invocation_results_fixed8.csv",
    "invocation_results_fixed9.csv",
    "invocation_results_fixed10.csv"
]
kourier = [
   "kourier.csv"
]

pods = [
    "logs/containers_live_20260415_094146.csv",
    "logs/containers_live_20260415_104138.csv",
    "logs/containers_live_20260415_105239.csv",
    "logs/containers_live_20260415_110513.csv",
    "logs/containers_live_20260415_111355.csv",
    "logs/containers_live_20260415_113705.csv",
    "logs/containers_live_20260415_114747.csv",
    "logs/containers_live_20260415_115937.csv"
  
   
]

auto = [
    
    "reconcile_times1.csv",
    "reconcile_times2.csv",
    "reconcile_times3.csv",
    "reconcile_times4.csv"
    
]

acti = [
    "context_handler_times1.csv",
    "context_handler_times2.csv",
    "context_handler_times3.csv",
    "context_handler_times4.csv",
    "context_handler_times5.csv",
    "context_handler_times6.csv",
    "context_handler_times7.csv",
   
    
]
scheduler = pd.read_csv('scheduler_times.csv')

# concaténation

#invocation
dfs = [pd.read_csv(f) for f in files]
all_data = pd.concat(dfs, ignore_index=True)

#kourier
kourier_dfs = [pd.read_csv(f) for f in kourier]
kourier_df = pd.concat(kourier_dfs, ignore_index=True)

#pod queue-proxy
dfs = [pd.read_csv(f) for f in pods]  
pod_df = pd.concat(dfs, ignore_index=True)  

#autoscaler
dfs = [pd.read_csv(f) for f in auto]  
auto_df = pd.concat(dfs, ignore_index=True)  

#activator
dfs = [pd.read_csv(f) for f in acti]  
acti_df = pd.concat(dfs, ignore_index=True)  

#acti = pd.read_csv('context_handler_times1.csv')
#auto = pd.read_csv('reconcile_times1.csv')
#df_percentil = pd.read_csv('latency_percentiles.csv')
```

```python
#conversion en ms

all_data['execution_time_sec'] = all_data['execution_time_sec'] * 1000


pod_df["createdAt"] = pd.to_datetime(pod_df["createdAt"], errors="coerce", utc=True)
pod_df["startedAt"] = pd.to_datetime(pod_df["startedAt"], errors="coerce", utc=True)
pod_df = pod_df.dropna(subset=["createdAt", "startedAt"])

# Vérification
print(pod_df.dtypes[["createdAt", "startedAt"]])

pod_df["duration_ms"] = (pod_df["startedAt"] - pod_df["createdAt"]).dt.total_seconds() * 1000
# Vérifier le résultat
print(pod_df[["pod", "container_name", "duration_ms"]].head())
pod_df.to_csv("pod_df_200ms.csv", index=False)
```

```python
all_data.to_csv("all_data_200ms.csv", index=False)
acti_df.to_csv("acti_df_200ms.csv", index=False)
auto_df.to_csv("auto_df_200ms.csv", index=False)
kourier_df.to_csv("acti_df_200ms.csv", index=False)
```

```python

# Récupérer les durées
durations = scheduler["scheduling_time_ms"].dropna().values

# Trier les valeurs
durations_sorted = np.sort(durations)

# Calcul CDF
cdf = np.arange(1, len(durations_sorted)+1) / len(durations_sorted)

# Plot
plt.figure(figsize=(8,6))
plt.plot(durations_sorted, cdf)

plt.xlabel("Scheduling Latency (ms)")
plt.ylabel("CDF")
#plt.title("CDF of Scheduler Durations")
plt.grid(True)
plt.savefig("scheduler.pdf", format="pdf") 
plt.show()
```

# Queue_proxy

```python

plt.figure(figsize=(7,5))


sns.ecdfplot(data=pod_df, x="duration_ms",label="Queue-proxy")

plt.xlabel("Latency Queue-proxy (ms)")
plt.ylabel("CDF")
#plt.title("ECDF Comparison of Latencies")
plt.legend()

plt.tight_layout()
plt.savefig("pod_2ms.pdf", format="pdf")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Exemple : pod_df contient les durées individuelles
# pod_df['duration_ms'] = latences de chaque invocation

# Calcul des percentiles
p50 = pod_df['duration_ms'].quantile(0.50)
p95 = pod_df['duration_ms'].quantile(0.95)
p99 = pod_df['duration_ms'].quantile(0.99)

plt.figure(figsize=(10,7))

# Courbe des durées individuelles
plt.plot(pod_df.index, pod_df['duration_ms'], alpha=0.7)

# Ajouter les percentiles comme lignes horizontales
plt.axhline(p50, color='green', linestyle='--', linewidth=2, label='p50')
plt.axhline(p95, color='orange', linestyle='--', linewidth=2, label='p95')
plt.axhline(p99, color='red', linestyle='--', linewidth=2, label='p99')

plt.axhline(0, color='black', linewidth=1)  # ligne de référence
plt.xlabel("Sandbox count ")
plt.ylabel("Latency (ms)")
plt.legend()
plt.grid(True)
plt.savefig("Queue_proxy.pdf", format="pdf")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,7))

# Boxplot
sns.boxplot(
    x='percentile',
    y='latency_s',
    hue='count_pod',
    data=df_long,
    width=0.6,
    dodge=True,
    palette="Set2"  # couleurs distinctes par count_pod
)

# Points individuels
sns.stripplot(
    x='percentile',
    y='latency_s',
    hue='count_pod',
    data=df_long,
    dodge=True,
    jitter=0.25,   # dispersion horizontale
    alpha=0.8,
    size=6,        # points plus gros
    palette="Set2",
    linewidth=0
)

# Supprimer la légende supplémentaire créée par stripplot
plt.legend(title='count_pod', bbox_to_anchor=(1.05,1), loc='upper left')

plt.xlabel("Percentile")
plt.ylabel("Latence (s)")
plt.title("Latence selon les percentiles et le nombre de pods")
plt.grid(True)
plt.tight_layout()
plt.show()
```
